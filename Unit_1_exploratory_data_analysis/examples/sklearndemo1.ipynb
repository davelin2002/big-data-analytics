# Default Datasets

```python
from sklearn.datasets import load_iris

import pandas as pd

import matplotlib.pyplot as plt

iris_bunch = load_iris(as_frame=True)

df = iris_bunch.frame

# display(df.shape)
# display(df.head)

#df.groupby('target').mean()

iris_filtered_setosa = df[df['target']==0]

iris_filtered_veriscolor = df[df['target']==1]

iris_filtered_virginica = df[df['target']==2]

filtered_iris_mean = [iris_filtered_setosa.mean(), 
     iris_filtered_veriscolor.mean(), 
     iris_filtered_virginica.mean()]

iris_mean = pd.concat(filtered_iris_mean, axis = 1)

# standard deviation stdev
# var var
# median median

display(iris_mean)

display(df.groupby('target').std())

display(df.groupby('target').var())

display(df.groupby('target').median())

display(df.groupby('target').describe())


iris_mean.plot(kind = 'bar')














```
