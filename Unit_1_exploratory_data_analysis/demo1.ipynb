Each data pipleline follows similar steps being:

6. clean up

## Imports and Loading CSV as Dataframe

```python
import pandas as pd

df = pd.read_csv('taxi_data_subset.csv')
```

## Observe the shape of the dataframe and data types

```python
display(df.shape)

print()

display(df.dtypes)

print()

display(df.dtypes[[
    'tpep_pickup_datetime',
    'fare_amount',
    'tip_amount'
]])


```

```python
display(df.head())

print()


display(df.tail())
```

## Parse the pickup date time

```python
df['tpep_pickup_datetime'] = pd.to_datetime(
    df['tpep_pickup_datetime']
)

display(df['tpep_pickup_datetime'])
```

## extract the hours

```python
df['hour'] = df['tpep_pickup_datetime'].dt.hour

display(df['hour'])
```

## group by hour and average the tip

```python
avg_hourly_tip = df.groupby('hour')['tip_amount'].mean()

display(avg_hourly_tip)
```

## clean up the results

```python
avg_hourly_tip = avg_hourly_tip.round(2)

avg_hourly_tip = avg_hourly_tip.sort_index()

display(avg_hourly_tip)
```
